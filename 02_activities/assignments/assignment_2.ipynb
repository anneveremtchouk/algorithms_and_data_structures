```python
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right

from collections import deque

def build_tree(levels):
    """
    Builds a binary tree from a level-order list.
    Example of a level-order list: [1, 10, 2, 3, 10, 12, 12]
    Returns the root node of the constructed binary tree.
    """
    if not levels:
        return None

    root = TreeNode(levels[0])
    queue = deque([root])
    i = 1  # Index to track position in 'levels'
    
    # Use BFS to link children
    while queue and i < len(levels):
        node = queue.popleft()
        
        # Left child
        if i < len(levels) and levels[i] is not None:
            node.left = TreeNode(levels[i])
            queue.append(node.left)
        i += 1
        
        # Right child
        if i < len(levels) and levels[i] is not None:
            node.right = TreeNode(levels[i])
            queue.append(node.right)
        i += 1
    
    return root

def is_duplicate(root: TreeNode) -> int:
  # TODO
  
  if not root:
        return -1

  seen = set()
  queue = deque([root])
    
  while queue:
    node = queue.popleft()
    # Check if the node's value is already seen.
    if node.val in seen:
        return node.val
    seen.add(node.val)
    # Add children to the queue if they exist.
    if node.left:
        queue.append(node.left)
    if node.right:
        queue.append(node.right)

  # No duplicate was found.
  return -1
```
