  # Question Two: Path to Leaves

  Given the `root` of a binary tree, return all root to leaf paths in any order.

  ## Examples

  ### Example 1

  ![](./images/q1_ex1.png)

  Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

  ### Example 2

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 7, 8]`

  Output: [[10, 7], [10, 9, 8]]

</details>

#### Starter Code for Question 2

```python
def bt_path(root):
    # if root is None > return (if two lines below are missing > running on None results in error shown)
    if root is None:
        return
    
    # assign queue to root. assign root node value to branch 
    # create empty list complete_tree that will store all branches 
    queue = [root]
    branch = [[root.val]]
    complete_tree = []
    
    while queue:
        
        # retrieve element from the queue and from branch with index 0
        # we retirving here root and root node value at the first cycle of function
        current_node = queue.pop(0)
        current_branch = branch.pop(0)
        
        # this is for leaf nodes, it will be the end of the while loop
        if current_node.left is None and current_node.right is None:
            complete_tree.append(current_branch)
        
        # to catch nodes on the left edges
        if current_node.left: 
            queue.append(current_node.left)
            branch.append(current_branch+[current_node.left.val])
        # to catch nodes  on the right edges
        if current_node.right:
            queue.append(current_node.right)
            branch.append(current_branch+[current_node.right.val])

    return complete_tree
```

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
```

```python
# Checking code on the example 1

root1 = TreeNode(1)
root1.left = TreeNode(2)
root1.right = TreeNode(2)
root1.left.left = TreeNode(3)
root1.left.right = TreeNode(5)
root1.right.left = TreeNode(6)
root1.right.right = TreeNode(7)
bt_path(root1)
```

```python
# Checking code on the example 2

root2 = TreeNode(10)
root2.left = TreeNode(9)
root2.right = TreeNode(7)
root2.left.left = TreeNode(8)
bt_path(root2)
```
